```python
import pandas as pd
import matplotlib.pyplot as plt

# Load your dataset
df = pd.read_csv('../data/articles_annotated.csv')

# Calculate percentage distribution
code_percentages = df['coding'].value_counts(normalize=True) * 100

# Plot a bar chart with default colors
plt.figure(figsize=(10, 6))
plt.bar(code_percentages.index, code_percentages.values)

# Add labels and title
plt.title('Percentage Distribution of Coding Categories in News Articles')
plt.xlabel('Coding Categories')
plt.ylabel('Percentage (%)')
plt.xticks(rotation=45)
plt.show()
```
